# UniChart at scale — large datasets, deltas, and the full formatting toolkit

This notebook is a **guided, self-checking tour** of UniChart's data engine and plotting
API on realistically large data. Every section ends with `assert`s, so the whole notebook
doubles as a regression suite — if it runs to the final `ALL SECTIONS PASSED`, the
behaviors it demonstrates are intact.

**What it covers**

| Part | Sections | Theme |
|---|---|---|
| I | 1–4 | The data engine: batched loading, cheap column access, own-column views |
| II | 5–6 | Plotting wide steady-state data and 60k-row transients (WebGL) |
| III | 7–12 | The formatting toolkit: per-set styling, hue, regressions, decorations, multi-axis, per-variable formats |
| IV | 13–15 | Bars, boxes, histograms, contours with overlays, queries |
| V | 16–18 | Deltas: row-matched and interpolated, then styled delta plots & tables |
| VI | 19 | Combining sets, resets, and the final wrap-up checks |

**Why it's fast.** UniChart keeps every set's rows in **one combined DataFrame** tagged by
`_SET_ID`. The engine layers four things on top so that width never hurts you:
a cached per-set **row-position index** (no full-column scans), **narrow column access**
(`ds.cols(...)` — plots fetch only the columns they draw, *then* sort), **batched
registration** (a multi-set file is one concat, not one per set), and **own-column
tracking** (each set's `df` shows only the columns it owns, so other sets' all-NaN
"phantom" columns never widen your slices). Scatter traces above
`unichart.WEBGL_POINT_THRESHOLD` points render with WebGL (`Scattergl`) instead of SVG.

Requirements: `pandas`, `numpy`, `plotly`, `scipy`, `ipywidgets`.

## 1. Build a realistic large dataset

We simulate a jet-engine test campaign:

- **Steady-state**: 4 engine builds × 1,500 stabilized points, with 8 real performance
  parameters (`FN` thrust, `T4`, `EGT`, `WF`, `N1`, `N2`, `SFC`, `PHASE`) **plus 2,000
  padding sensor channels** (`CH0000`–`CH1999`) — the classic "wide file" where most tools
  bog down. One long-format frame, split by `SETNUMBER`.
- **Transient**: a single 60,000-row recording (100 Hz for 10 minutes) of an accel/decel.
- **Envelope**: an 80-point operating-boundary trace we'll lay over contour plots.

```python
# --- make repo-root importable (notebook lives in demo_notebooks/) ---
import sys, os
_repo_root = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
if _repo_root not in sys.path:
    sys.path.insert(0, _repo_root)

import time
import numpy as np
import pandas as pd
import plotly.graph_objects as go

import unichart
from unichart import UnichartNotebook

rng = np.random.default_rng(42)
tic = time.perf_counter

# ---- steady-state: 4 builds x 1,500 rows x (8 + 2,000) columns -------------
N_PAD, ROWS_PER_SET, BUILDS = 2000, 1500, 4
TITLES = ['Build A (baseline)', 'Build B (nozzle mod)',
          'Build C (new HPC)', 'Build D (combustor tweak)']

frames = []
for b in range(BUILDS):
    fn   = np.sort(rng.uniform(20, 120, ROWS_PER_SET))            # thrust sweep [kN]
    bias = [0.0, 12.0, -18.0, 6.0][b]                              # per-build shift
    t4   = 850 + 6.5*fn + 0.012*fn**2 + bias + rng.normal(0, 6, ROWS_PER_SET)
    egt  = 0.42*t4 + 180 + 0.4*bias + rng.normal(0, 4, ROWS_PER_SET)
    wf   = 0.085*fn**1.18 + 0.4 + 0.002*bias + rng.normal(0, 0.05, ROWS_PER_SET)
    n1   = 28*np.log(fn) - 30 + rng.normal(0, 0.6, ROWS_PER_SET)
    frames.append(pd.DataFrame({
        'SETNUMBER': b + 1, 'TITLE': TITLES[b],
        'FN': fn, 'T4': t4, 'EGT': egt, 'WF': wf,
        'N1': n1, 'N2': 0.62*n1 + 52 + rng.normal(0, 0.4, ROWS_PER_SET),
        'SFC': wf/fn*1000,
        'PHASE': np.where(fn < 50, 'idle', np.where(fn < 90, 'cruise', 'takeoff')),
    }))
core = pd.concat(frames, ignore_index=True)
pad  = pd.DataFrame(rng.standard_normal((len(core), N_PAD)),
                    columns=[f'CH{i:04d}' for i in range(N_PAD)])
steady = pd.concat([core, pad], axis=1)

# ---- transient: one 60,000-row accel/decel recording -----------------------
t = np.arange(0, 600, 0.01)
n1_t = 45 + 35/(1 + np.exp(-(t - 120)/18)) - 22/(1 + np.exp(-(t - 420)/25))
transient = pd.DataFrame({
    'TIME':  t,
    'N1':    n1_t + rng.normal(0, 0.15, t.size),
    'N2':    0.62*n1_t + 52 + rng.normal(0, 0.10, t.size),
    'EGT_T': 9.5*n1_t + 120 + rng.normal(0, 3.0, t.size),
    'WF_T':  0.02*n1_t**1.5 + rng.normal(0, 0.02, t.size),
})

# ---- operating-envelope boundary (drawn over contours later) ---------------
ang = np.linspace(0, 2*np.pi, 80)
envelope = pd.DataFrame({'FN': 70 + 48*np.cos(ang),
                         'WF': 13 + 11*np.sin(ang),
                         'ORDER': np.arange(ang.size, dtype=float)})

print(f"steady    : {steady.shape[0]:,} rows x {steady.shape[1]:,} cols "
      f"(~{steady.memory_usage(deep=False).sum()/1e6:,.0f} MB)")
print(f"transient : {transient.shape[0]:,} rows x {transient.shape[1]} cols")
print(f"envelope  : {envelope.shape[0]} boundary points")
```

## 2. Loading: one batched rebuild, not one per set

`load()` accepts DataFrames, dicts, file paths (`.csv/.tsv/.xlsx/.json/.parquet`...),
or numpy arrays. A frame carrying a `SETNUMBER` (or `INDEX`) column is split into one set
per distinct value — and all of those sets are registered with a **single** combined-frame
concat. (Registering one at a time would re-copy the whole accumulated frame per set,
which is quadratic in the number of sets.)

```python
nb = UnichartNotebook()

t0 = tic()
nb.load(steady)                       # SETNUMBER -> 4 sets, ONE combined-frame build
t_load = tic() - t0
print(f"\nLoaded {len(nb.sets)} sets from one {steady.shape[1]:,}-column frame "
      f"in {t_load:.2f}s (single batched concat)\n")

nb.list_sets()

assert len(nb.sets) == 4
assert [d.title for d in nb.sets] == TITLES
assert nb.df.shape == (BUILDS*ROWS_PER_SET, steady.shape[1] + 1)   # +_SET_ID tag
assert t_load < 30
print('\nPASS - multi-set wide file loaded in one batched rebuild.')
```

```python
# Files work the same way; the title defaults to the filename stem.
import tempfile, os
tmp = tempfile.mkdtemp()
csv_path = os.path.join(tmp, 'leak_check_run.csv')
core.head(200).drop(columns=['SETNUMBER', 'TITLE']).to_csv(csv_path, index=False)

nb_file = UnichartNotebook()
nb_file.load(csv_path)
assert nb_file.sets[0].title == 'leak_check_run'
os.remove(csv_path)
print('PASS - file loading works; title taken from the filename stem.')
```

## 3. Cheap access: read 3 columns, pay for 3 columns

`ds['col']` and `ds.cols([...])` copy **only the requested columns** of the set's rows
(found via a cached row-position index). `ds.df` still materializes the full set when you
genuinely want it — but nothing else on the hot path does. The column statistics
(`mean/max/min/median`) and `summary()` ride the same narrow path.

```python
ds0 = nb.sets[0]

t0 = tic(); egt  = ds0['EGT'];                  t_col  = tic() - t0
t0 = tic(); few  = ds0.cols(['FN', 'T4', 'EGT']); t_cols = tic() - t0
t0 = tic(); full = ds0.df;                      t_full = tic() - t0

print(f"ds['EGT']             : {t_col*1000:8.2f} ms   ({len(egt):,} values)")
print(f"ds.cols([3 columns])  : {t_cols*1000:8.2f} ms   {few.shape}")
print(f"ds.df (full width)    : {t_full*1000:8.2f} ms   {full.shape}   <- pay only when needed")

t0 = tic()
stats = (nb.mean('all', 'EGT'), nb.max('all', 'T4'),
         nb.min(0, 'FN'), nb.median([1, 2], 'SFC'))
print(f"4 cross-set aggregates: {(tic()-t0)*1000:8.2f} ms   -> "
      + ', '.join(f'{s:,.2f}' for s in stats))

assert np.isclose(egt.mean(), full['EGT'].mean())
assert list(few.columns) == ['FN', 'T4', 'EGT']
assert t_col < 0.5 and t_cols < 0.5

print()
_ = nb.summary(cols=['FN', 'T4', 'EGT', 'SFC'], print_table=True)
print('\nPASS - column reads no longer materialize the full set width.')
```

## 4. Own columns: union storage, owned views

All sets share one frame, so *storage* is the union of every set's columns (NaN-filled
where a set has no values). But each set's `df`/`columns` shows **only the columns it
owns** — so loading a 2,000-channel steady file next to a transient recording doesn't
pollute either view with the other's columns.

Ownership follows the data:
- any **API write** (`ds['col'] = ...`, `set_column`, `add_column`) claims the column for
  the written set(s) — including filling NaNs into a column some other set introduced;
- columns **added/dropped directly** on the live `nb.df` are reconciled automatically
  (owners assigned from the data);
- the one blind spot — in-place `.loc` value surgery on the live frame — is healed by
  `nb.refresh_own_columns(rescan=True)`.

```python
nb.load_df(transient, title='Transient recording')
ds_tr = nb.sets[-1]
nb.load_df(envelope, title='Operating envelope')
ds_env = nb.sets[-1]

nb.list_sets()      # the Shape column shows each set's OWN width

# Views are owned...
assert 'TIME' not in nb.sets[0].columns          # steady sets don't see transient channels
assert 'CH0000' not in ds_tr.columns             # and vice versa
assert list(ds_env.columns) == ['FN', 'WF', 'ORDER']
# ...while storage is the shared union:
assert 'TIME' in nb.df.columns

# (1) API writes claim columns for the written set(s) only
ds0['EGT_MARGIN'] = 950 - ds0['EGT']
assert 'EGT_MARGIN' in ds0.columns
assert 'EGT_MARGIN' not in nb.sets[1].columns

# ...including NaN-filling a column another set introduced:
vals = 950 - pd.concat([nb.sets[i]['EGT'] for i in (1, 2, 3)]).to_numpy()
nb.set_column([1, 2, 3], 'EGT_MARGIN', vals)
assert all('EGT_MARGIN' in nb.sets[i].columns for i in range(4))
assert 'EGT_MARGIN' not in ds_tr.columns         # untouched sets stay clean

# (2) direct column adds on nb.df are reconciled from the data automatically
flag = pd.Series(np.nan, index=nb.df.index)
flag.iloc[nb._set_positions(ds_tr._set_id)] = 1.0
nb.df['IS_TRANSIENT'] = flag
assert 'IS_TRANSIENT' in ds_tr.columns           # owner found from the data
assert 'IS_TRANSIENT' not in ds0.columns

# (3) in-place .loc surgery is invisible to bookkeeping -> heal with a rescan
pos = nb._set_positions(ds_env._set_id)
nb.df.loc[nb.df.index[pos], 'EGT_MARGIN'] = 0.0
assert 'EGT_MARGIN' not in ds_env.columns
nb.refresh_own_columns(rescan=True)
assert 'EGT_MARGIN' in ds_env.columns
print('PASS - owned views; API claiming; auto-reconcile; rescan healing.')
```

## 5. Plotting wide data

`plot()` resolves which columns each subplot needs **first**, fetches that narrow slice,
sorts it, and only then builds traces — so a 4-subplot scatter against a 2,000-channel
frame costs milliseconds, not a full-width sort per set. Hover tooltips
(`set_display_parms`) ride in a slim `customdata` array (x/y are not duplicated into it).

Also on display here: case-insensitive x lookup (`x='fn'`), `subplot_titles`,
`figsize`, `ncols`, and a persistent `suptitle`.

```python
nb.select([0, 1, 2, 3])                              # steady sets only
nb.set_display_parms('all', ['EGT', 'SFC', 'PHASE']) # hover parms
nb.suptitle = 'Steady-state performance - 4 builds'

t0 = tic()
fig = nb.plot(x='fn', y=['T4', 'EGT', 'SFC', 'N1'], ncols=2,   # note: lowercase 'fn' resolves
              subplot_titles=['Turbine temp', 'Exhaust temp', 'Spec. fuel cons.', 'Fan speed'])
t_plot = tic() - t0
print(f"4-subplot scatter across a {nb.df.shape[1]:,}-column frame: {t_plot*1000:,.0f} ms")

assert t_plot < 10
assert len(fig.data) == 16                                    # 4 sets x 4 pairs
assert np.asarray(fig.data[0].customdata).shape[1] == 3       # hover parms only
fig
```

## 6. Transients: 60,000 points per trace, rendered with WebGL

SVG scatter creates one DOM node per point and locks the browser on dense data. Above
`unichart.WEBGL_POINT_THRESHOLD` points (default 10,000) traces switch to `go.Scattergl`.
Set the threshold to `None` to force SVG everywhere.

```python
nb.select(ds_tr)
nb.set_display_parms(ds_tr, ['N2'])
nb.marker(ds_tr, None)                 # line-only rendering for the recording
nb.linestyle(ds_tr, '-')
nb.linewidth(ds_tr, 1)
nb.color(ds_tr, '#16a085')
ds_tr.order = 'TIME'                   # connect points in time order

fig = nb.plot(x='TIME', y=['N1', 'EGT_T'], ncols=1, nrows=2,
              suptitle='Accel/decel transient - 60k samples per trace')

assert all(isinstance(tr, go.Scattergl) for tr in fig.data)      # WebGL above threshold
assert np.asarray(fig.data[0].customdata).shape == (len(transient), 1)

# The threshold is a module-level switch:
unichart.WEBGL_POINT_THRESHOLD = None
fig_svg = nb.plot(x='TIME', y='N1', ncols=1, nrows=1)
assert isinstance(fig_svg.data[0], go.Scatter)
unichart.WEBGL_POINT_THRESHOLD = 10_000
print('PASS - Scattergl above the threshold; plain Scatter when disabled.')
fig
```

## 7. Formatting I — per-set styling

Everything a set draws with is a per-set attribute, settable through the notebook-level
helpers (which all take the usual selector: an index, a list, a `Dataset`, or `'all'`):

`color` (names / hex / palette-slot ints), `marker` (matplotlib or plotly names),
`markersize`, `linestyle`, `linewidth`, `edgewidth`, `fill` (hollow markers), `alpha`,
plus `set_title` and `set_color_palette`. The default assignment maps (`nb.color_map`,
`nb.marker_map`) are user-replaceable cyclic lists used at load/reset time.

```python
nb.select([0, 1, 2, 3])

# Custom default maps: what new/reset sets receive, by index (indexing cycles).
nb.color_map  = ['#0072B2', '#E69F00', '#009E73', '#CC79A7']
nb.marker_map = ['o', 's', 'D', '^']
nb.reset_format(uset_slice=[0, 1, 2, 3], vars=False, lines=False,
                highlights=False, scale=False, fonts=False)
assert nb.sets[1].color == '#E69F00' and nb.sets[1].marker == 's'
assert nb.color_map[7] == nb.color_map[3]            # CyclicList wraps

# Targeted overrides on top of the map defaults:
nb.color(3, 'crimson')                # color name
nb.color(2, '#7f7f7f')                # hex
nb.color(1, 0)                        # int -> the color_map color of slot 0
nb.marker(3, 'triangle-up')           # plotly-native marker names pass through
nb.markersize('all', 7)
nb.linestyle([0, 1], '-')
nb.linestyle(2, '--')
nb.linestyle(3, 'dashdot')
nb.linewidth('all', 1.5)
nb.edgewidth(3, 2)
nb.fill(2, 'off')                     # hollow markers, outline takes the set color
nb.alpha('all', 0.85)
nb.set_title(3, 'Build D (combustor tweak, rev2)')

# Dataset-level dict round-trip:
fmt = nb.sets[0].get_format_dict()
assert fmt['marker'] == 'o' and fmt['linestyle'] == '-'
nb.sets[0].update_format_dict({'markersize': 9})
assert nb.sets[0].markersize == 9

fig = nb.plot(x='FN', y='EGT', ncols=1, nrows=1, figsize=(11, 6),
              suptitle='Per-set styling: colors, markers, lines, hollow fill')
assert fig.data[0].marker.size == 9
assert fig.data[2].marker.color == 'rgba(0,0,0,0)'   # set 2 is hollow
assert 'rev2' in fig.data[3].name
fig
```

## 8. Formatting II — hue (color-by-variable)

`hue(sets, column)` colors each *point* by a column. Numeric hues share a figure-level
coloraxis with a colorbar (range pinnable via `scale`); categorical hues map category
codes through the set's `hue_palette`.

```python
nb.hue([0, 1, 2, 3], 'EGT_MARGIN')          # numeric hue -> shared colorbar
nb.hue_palette('all', 'Viridis')
nb.scale('EGT_MARGIN', (0, 350))             # pins the colorbar range too

fig = nb.plot(x='FN', y='T4', ncols=1, nrows=1,
              suptitle='Numeric hue: EGT margin, pinned 0-350')
assert fig.layout.coloraxis.colorscale is not None
assert (fig.layout.coloraxis.cmin, fig.layout.coloraxis.cmax) == (0, 350)
nb.scale('EGT_MARGIN', 'clear')
fig
```

```python
nb.hue('all', 'PHASE')                       # categorical hue
nb.hue_palette([0, 1, 2, 3], 'Portland')     # continuous scale; codes are mapped through it
fig = nb.plot(x='FN', y='SFC', ncols=1, nrows=1,
              suptitle='Categorical hue: flight phase')
assert len(fig.data) == 4
nb.hue('all', '')                            # clear hue
fig
```

## 9. Formatting III — regressions and `reg_info`

`reg_order(sets, spec)` overlays a fitted curve per set. Specs: polynomial degrees
(`1`, `'poly2'`, `'cubic'`), `'log'`, `'exp'`, `'power'`, `'spline'`, `'lowess'`,
or a moving average `('ma', window)`. `reg_info()` prints the fitted equation and
R²/RMSE/MAE for the last-plotted x/y.

```python
nb.reg_order(0, 1)              # linear
nb.reg_order(1, 'poly2')        # quadratic
nb.reg_order(2, ('ma', 60))     # centered moving average
nb.reg_order(3, 'power')        # y = A x^b

fig = nb.plot(x='FN', y='T4', ncols=1, nrows=1,
              suptitle='Per-set regression overlays')
assert len(fig.data) == 8                      # 4 data + 4 fit traces
assert sum('Fit (' in (tr.name or '') for tr in fig.data) == 4

info = nb.reg_info()
assert info[1]['label'].startswith('Polynomial')
assert info[1]['stats']['r2'] > 0.99           # T4(FN) is genuinely quadratic
nb.reg_order('all', None)
fig
```

## 10. Formatting IV — reference lines, highlight bands, axis limits, fonts, dark mode

Decorations are keyed by *column name* and persist across plots: a line on the x-variable
draws vertical, on a y-variable horizontal; `highlight` shades a band; `scale` pins axis
limits (accepts a list of columns). `set_font_sizes` takes numbers or named sizes
(`'sm'`, `'lg'`, `'xl'`, ...), and `toggle_darkmode` flips the template.

```python
nb.line('FN', 90, color='orange', linestyle='--')          # vertical @ takeoff thrust
nb.line('EGT', 900, color='red', linestyle='-.')           # horizontal EGT redline
nb.highlight('FN', (45, 55), color='steelblue', alpha=0.15)
nb.highlight('EGT', (880, 920), color='red', alpha=0.10)
nb.scale('EGT', (550, 1000))
nb.scale(['FN'], (15, 125))
nb.set_font_sizes(suptitle='xl', axes_title=13, axes_tick='sm',
                  legend='sm', subplot_title='lg', hover='md')
nb.toggle_darkmode(True)

fig = nb.plot(x='FN', y=['EGT', 'T4'], ncols=2,
              suptitle='Decorated dark-mode view')
assert len(fig.layout.shapes) >= 4                 # lines + bands landed
assert tuple(fig.layout.yaxis.range) == (550, 1000)
assert fig.layout.title.font.size == 18            # 'xl'

nb.toggle_darkmode(False)
fig
```

```python
# Decorations persist until cleared:
nb.line('all', 'clear')
nb.highlight('all', 'clear')
nb.scale(['EGT', 'FN'], 'clear')
nb.set_font_sizes(reset=True)
assert nb.lines == {} and nb.highlights == {} and nb.axis_limits == {}
assert nb.get_font_sizes()['suptitle'] is None
print('PASS - decorations and fonts cleared.')
```

## 11. Formatting V — multi-Y-axis plots and per-variable overrides

`plot_ymult` overlays several y-variables on one x-axis, each with its own y-axis.
`var_format(variable, ...)` sets **per-variable** styling that beats the per-set styling
attribute-by-attribute — so "T4 is always red and solid, whatever set it comes from"
is one line. (The notebook-level `color`/`marker`/`linestyle`/... helpers auto-detect a
variable name as the target, too.)

```python
nb.var_format('T4', color='#d62728', linestyle='-', markersize=5)
nb.var_format('WF', color='#2ca02c', linestyle='--', marker='s')
nb.color('N1', '#9467bd')             # helper auto-detects a variable target
nb.linestyle('N1', ':')
nb.list_var_formats()

fig = nb.plot_ymult(x='FN', y=['T4', 'WF', 'N1'],
                    legend='above', legend_group_by='vars',
                    suptitle='Three y-axes, per-variable styling')
assert fig.layout.yaxis2.overlaying == 'y'
assert fig.layout.yaxis3 is not None
assert len(fig.data) == 12                       # 4 sets x 3 variables

nb.var_format('N1', color='reset')               # clear one attribute
nb.clear_var_format('WF')                        # clear one variable
nb.clear_var_format()                            # clear everything
assert nb.variable_formats == {}
fig
```

## 12. Formatting VI — per-dataset subplots and legend suppression

`by='sets'` gives each dataset its own subplot; with two y-variables the second gets a
secondary axis in every panel. `suppress_legends=True` starts every trace in
"legend-only" mode (click to reveal) — handy for crowded comparisons.

```python
fig = nb.plot(x='FN', y=['T4', 'SFC'], by='sets',
              suptitle='One panel per build, secondary axis for SFC')
assert len(fig.data) == 8                        # 4 sets x 2 vars

fig2 = nb.plot(x='FN', y='EGT', ncols=1, nrows=1, suppress_legends=True)
assert all(tr.visible == 'legendonly' for tr in fig2.data)
fig
```

## 13. Bars, boxes, and histograms

The same notebook front-end drives bar charts (`bar`), box plots (`box`), and histograms
(`histogram`), each with `by='vars'` (subplot per variable), `by='sets'` (subplot per
dataset), and — for bars/boxes — `by='dataset_x'` (datasets as the x-axis, one
independently-scaled y-axis per variable). Bar charts take `markers=` overlays (e.g.
limits), styled via `var_format`.

```python
# A compact per-rating table makes crisp bars: 3 builds x 4 ratings.
rt = []
for b in range(3):
    rt.append(pd.DataFrame({
        'SETNUMBER': b, 'TITLE': TITLES[b],
        'RATING':   ['Idle', 'Cruise', 'MCT', 'Takeoff'],
        'EGT_MEAN': np.array([405., 720., 870., 925.]) + 9*b + rng.normal(0, 3, 4),
        'T4_MEAN':  np.array([990., 1480., 1690., 1790.]) + 14*b + rng.normal(0, 5, 4),
        'EGT_LIMIT': 950.0,
    }))
nb2 = UnichartNotebook()
nb2.load_df(pd.concat(rt, ignore_index=True))
nb2.var_format('EGT_LIMIT', color='red', marker='*', markersize=16)

fig = nb2.bar(x='RATING', y='EGT_MEAN', markers='EGT_LIMIT', barmode='group')
assert sum(isinstance(tr, go.Bar) for tr in fig.data) == 3        # one bar series per build
assert sum(isinstance(tr, go.Scatter) for tr in fig.data) == 3    # limit markers per build
fig
```

```python
fig = nb2.bar(y=['EGT_MEAN', 'T4_MEAN'], by='dataset_x', agg='max')
assert len(fig.data) == 2 and fig.layout.yaxis2.side == 'right'
fig
```

```python
nb.select([0, 1, 2, 3])
fig = nb.box(x='PHASE', y=['EGT', 'SFC'], notched=True, points='outliers',
             ncols=2, suptitle='Distribution by phase')
assert sum(isinstance(tr, go.Box) for tr in fig.data) == 8
fig
```

```python
fig = nb.box(y=['EGT', 'T4'], by='dataset_x', boxmode='group')
assert fig.layout.yaxis2.overlaying == 'y'
fig
```

```python
fig = nb.histogram(x=['EGT', 'SFC'], nbins=40, barmode='overlay', alpha=0.55,
                   ncols=2, suptitle='Distributions, overlaid')
assert sum(isinstance(tr, go.Histogram) for tr in fig.data) == 8
fig
```

```python
# Weighted histogram with explicit bins, one panel per build:
fig = nb.histogram(x='EGT', y='FN', histfunc='avg', by='sets',
                   bin_size=25, bin_start=550, bin_end=1000,
                   suptitle='Mean thrust per EGT band')
assert all(tr.histfunc == 'avg' for tr in fig.data)
fig
```

## 14. Contours with styled overlays

`contour(x, y, z)` grids scattered data and draws filled contours; `overlay_sets=` lays
other datasets on top — honoring their full style. A set with a `linestyle` draws as a
connected line in its own `order`, perfect for an operating envelope. Overlays are
fetched and styled **once** and stamped onto every panel.

## 15. Queries and selection

`query(sets, expr)` masks rows non-destructively (pandas query syntax, strings included);
`query(sets, None)` restores them. A query that empties a set switches it off rather than
erroring. `select`/`omit`/`restore` flags drive every downstream operation.

```python
nb.select([0, 1, 2, 3])
nb.query([0, 1, 2, 3], 'PHASE == "cruise" and FN > 55')
_ = nb.summary(cols=['FN', 'EGT'], print_table=True)
assert all(0 < len(nb.sets[i].df) < ROWS_PER_SET for i in range(4))

fig = nb.plot(x='FN', y='EGT', ncols=1, nrows=1,
              suptitle='Cruise points above 55 kN only')
nb.query([0, 1, 2, 3], None)                       # masks gone, rows back
assert all(len(nb.sets[i].df) == ROWS_PER_SET for i in range(4))

nb.sets[0].sel_query('FN > 100')                   # per-set query setter
assert 0 < len(nb.sets[0].df) < ROWS_PER_SET
nb.sets[0].query = None
print('PASS - queries mask and restore rows without touching the frame.')
fig
```

## 16. Deltas I — row-matched (`merge_asof`) mode

`delta(base, studies, align_on=...)` builds, per study set, a new dataset anchored on the
**base** rows: every base column is carried through, and each delta parameter `P` gains
`DL_P` (study − base) and `DLPCT_P` (%). Options:

- `keep_parms` — also carry the raw study value as `P_STUDY`;
- `passed_parms` — carry extra study context columns as `<name>_STUDY`;
- `direction` / `tolerance` — control the nearest-row matching.

Thanks to own-column tracking, the result set owns exactly its columns — previous deltas'
`DL_*` outputs never bleed into the next delta or the original sets — and the empty-column
scan only spans the base set's true width.

```python
t0 = tic()
deltas = nb.delta(0, [1, 2, 3], align_on='FN',
                  delta_parms=['T4', 'EGT', 'SFC'],
                  keep_parms=['T4'], passed_parms='PHASE')
t_delta = tic() - t0
print(f"\n3 deltas against a {nb.df.shape[1]:,}-column frame in {t_delta:.2f}s")

assert len(deltas) == 3
d = deltas[0].df
dcols = list(d.columns)
for needed in ['FN', 'T4', 'T4_STUDY', 'DL_T4', 'DLPCT_T4', 'DL_EGT', 'DLPCT_EGT',
               'DL_SFC', 'DLPCT_SFC', 'PHASE_STUDY']:
    assert needed in dcols, needed
assert 'EGT_STUDY' not in dcols and 'SFC_STUDY' not in dcols   # keep_parms=['T4'] only
assert not any(d[c].isna().all() for c in dcols)         # no phantom/empty columns
assert 'DL_T4' not in nb.sets[0].columns                 # originals stay unpolluted
assert 'CH0000' in dcols                                 # base context carried through
# Build B ran hotter (+12), Build C cooler (-18) -> delta signs reflect it:
assert (deltas[0].df['DL_T4'].mean() > 0) and (deltas[1].df['DL_T4'].mean() < 0)
print('PASS - base-anchored deltas with kept study values and passthrough context.')
```

```python
# Matching is tunable: a tight tolerance leaves unmatched rows NaN (with a warning).
tight = nb.delta(0, 1, align_on='FN', delta_parms='T4',
                 direction='forward', tolerance=0.02)
frac = tight[0].df['DL_T4'].isna().mean()
print(f"direction='forward', tolerance=0.02  ->  {frac:.0%} of rows unmatched (NaN)")
assert frac > 0
```

## 17. Deltas II — interpolated (`x_ins`) mode

Pass `x_ins=[...]` to place the result rows at exact `align_on` values, reading each side
off a fitted curve (`kind=` accepts any `reg_order` spec) or from the nearest raw row.
`interp=` chooses which side(s) to interpolate, and `BASE_METHOD`/`STUDY_METHOD` record
how every value was produced (`'LS2'`, `'Table'`, `'Nearest'`, ...).

Here we compare builds **at rating thrusts** — study side on a quadratic fit, base side
nearest-row (which keeps the wide base cheap):

```python
ratings_fn = [40.0, 70.0, 100.0]
dx = nb.delta(0, 1, align_on='FN', delta_parms=['T4', 'SFC'],
              x_ins=ratings_fn, interp='study', kind='poly2',
              passed_parms='PHASE')
dd = dx[0].df
print(dd[['FN', 'DL_T4', 'DLPCT_T4', 'DL_SFC', 'PHASE_STUDY',
          'BASE_METHOD', 'STUDY_METHOD']].to_string(index=False))

assert list(dd['FN']) == ratings_fn
assert set(dd['STUDY_METHOD']) == {'LS2'}        # fitted curve
assert set(dd['BASE_METHOD'])  == {'Nearest'}    # nearest raw base row
assert len(dd) == len(ratings_fn)
```

```python
# With no spec at all, numeric columns interpolate piecewise-linearly ('Table'):
nb3 = UnichartNotebook()
xg = np.linspace(0, 10, 60)
nb3.load_df(pd.DataFrame({'X': xg, 'Y': 3*xg + 1 + rng.normal(0, .05, 60)}), title='base')
nb3.load_df(pd.DataFrame({'X': xg, 'Y': 3*xg + 4 + rng.normal(0, .05, 60)}), title='study')
db = nb3.delta(0, 1, align_on='X', delta_parms='Y', x_ins=[2.5, 5.0, 7.5], interp='both')
assert set(db[0].df['BASE_METHOD']) == {'Table'} == set(db[0].df['STUDY_METHOD'])
assert np.allclose(db[0].df['DL_Y'], 3.0, atol=0.2)
print('PASS - x_ins interpolation on both sides, METHOD columns label the source.')
```

## 18. Plotting deltas with the full toolkit

Delta sets are first-class: select them, restyle them, decorate the delta axes with a
zero line and a tolerance band, and read exact values back with `table()` —
raw rows, or interpolated at chosen thrusts with significant-figure rounding and
DataFrame/Markdown output.

```python
didx = [d.index for d in deltas]
nb.select(didx)
nb.set_color_palette('Dark2', didx)
nb.marker(didx, 'D')
nb.markersize(didx, 5)
nb.linestyle(didx, None)
nb.alpha(didx, 0.9)
nb.set_display_parms(didx, ['T4_STUDY', 'PHASE_STUDY'])

nb.line('DLPCT_T4', 0, color='gray', linestyle='-')          # zero reference
nb.highlight('DLPCT_T4', (-0.5, 0.5), color='green', alpha=0.12)  # +-0.5 % band
nb.suptitle = 'Build deltas vs Build A (baseline)'

fig = nb.plot(x='FN', y=['DL_T4', 'DLPCT_T4'], ncols=2)
assert len(fig.data) == 6                                    # 3 delta sets x 2 panels
assert len(fig.layout.shapes) >= 2
fig
```

```python
# Exact readback: raw table of one delta set...
nb.select(didx[0])
tbl = nb.table(cols=['FN', 'DL_T4', 'DLPCT_T4'], output='df')
assert {'Set', 'Dataset', 'FN', 'DL_T4', 'DLPCT_T4'} <= set(tbl.columns)
assert len(tbl) == ROWS_PER_SET

# ...and an interpolated lookup at rating thrusts, 3 significant figures, as Markdown:
md = nb.table(cols='DLPCT_T4', x_col='FN', x_in=[50, 75, 100],
              kind='poly2', sig_figs=3, output='md')
print(md)

# table_read() gives the raw interpolated numbers per set:
vals = nb.table_read(didx[0], 'FN', 'DL_T4', [60.0])
assert np.isfinite(vals[didx[0]]).all()
print('\nPASS - tables: raw, interpolated+sig-figs, markdown, and table_read.')
```
